```python
x = torch.tensor([[1, 2],
                   [3, 4],
                   [5, 6],
                   [7, 8]])  # (4, 2)

mask = torch.tensor([True, False, True, False])

x[mask]     # 筛行 → shape (2, 2) → [[1,2], [5,6]]
mask = torch.tensor([False, True])
x[:, mask]  # 筛列 → shape (4, 2的子集)，这里mask长度要匹配列数才行
```
